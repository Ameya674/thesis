# EDA

## What is the range of the dataset?

```python
start_date = acled["event_date"].min()
end_date   = acled["event_date"].max()

print("Dataset covers from",
      start_date.strftime("%d %B %Y"),
      "to",
      end_date.strftime("%d %B %Y"))
```

## How many incidents took place each year?

```python
counts_per_year = pd.to_datetime(acled["event_date"]).dt.year.value_counts().sort_index()
plt.figure(figsize=(10, 6))

bars = plt.bar(
    counts_per_year.index.astype(str),
    counts_per_year.values,
    color="#FF8C42",      # 🔸 new bar color
    edgecolor="black",
    alpha=0.9,
)

plt.title("Records per Year", fontsize=16, fontweight="bold", pad=15)
plt.xlabel("Year", fontsize=12)
plt.ylabel("Number of Records", fontsize=12)

plt.xticks(rotation=45, ha="right")
plt.grid(axis="y", linestyle="--", alpha=0.4)

for spine in ["top", "right"]:
    plt.gca().spines[spine].set_visible(False)

for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height,
        f"{int(height):,}",
        ha="center",
        va="bottom",
        fontsize=10,
        rotation=0,
    )

plt.tight_layout()
plt.show()
```

## What is the YoY increase/decrease of events for each year?

```python
yoy_pct_change = counts_per_year.pct_change()
yoy_pct_change = yoy_pct_change.dropna()

plt.figure(figsize=(10, 6))

# Green for increase, red for decrease
colors = ["#2E8B57" if v >= 0 else "#D62728" for v in yoy_pct_change.values]

bars = plt.bar(
    yoy_pct_change.index.astype(str),
    yoy_pct_change.values,
    color=colors,
    edgecolor="black",
    alpha=0.9,
)

plt.axhline(0, color="black", linewidth=1)

plt.title("Year-on-Year % Change in Incident Counts", fontsize=16, fontweight="bold", pad=15)
plt.xlabel("Year", fontsize=12)
plt.ylabel("Percentage Change (%)", fontsize=12)

plt.xticks(rotation=45, ha="right")
plt.grid(axis="y", linestyle="--", alpha=0.4)

for spine in ["top", "right"]:
    plt.gca().spines[spine].set_visible(False)

# Labels on bars
for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height,
        f"{height:.1f}%",                     # 1 decimal place
        ha="center",
        va="bottom" if height >= 0 else "top",
        fontsize=10,
    )

plt.tight_layout()
plt.show()
```

## What is the distribution of these events on a monthly basis?

```python
monthly_dataset = pd.DataFrame()
monthly_dataset["year"] = acled["event_date"].dt.year
monthly_dataset["month"] = acled["event_date"].dt.month
```

```python
monthly_dataset = (
    monthly_dataset.dropna(subset=["year", "month"])
                   .astype({"year": "int64", "month": "int64"})
)
```

```python
monthly_counts = (
    monthly_dataset.groupby(["year", "month"])
                   .size()
                   .rename("events")
                   .reset_index()
)
```

```python
monthly_counts
```

```python
# pivot as before
wide = (monthly_counts.pivot(index="year", columns="month", values="events")
                     .reindex(columns=range(1, 13), fill_value=0)
                     .fillna(0))

# pick a palette to rotate through
colors = ["#FF8C42", "#4C72B0", "#2E8B57", "#9467BD", "#E377C2"]

for i, (yr, row) in enumerate(wide.iterrows()):
    plt.figure(figsize=(10, 4))

    color = colors[i % len(colors)]  # change color for each year

    bars = plt.bar(
        range(1, 13),
        row.values,
        color=color,
        edgecolor="black",
        alpha=0.9,
    )

    ymax = row.values.max() if len(row.values) else 0
    plt.ylim(0, ymax * 1.10 if ymax > 0 else 1)

    plt.grid(axis="y", linestyle="--", alpha=0.4)
    for spine in ["top", "right"]:
        plt.gca().spines[spine].set_visible(False)

    # labels on bars
    for b in bars:
        h = b.get_height()
        if h > 0:
            plt.text(
                b.get_x() + b.get_width() / 2,
                h,
                f"{int(h)}",
                ha="center",
                va="bottom",
                fontsize=10,
            )

    plt.xticks(range(1, 13), [calendar.month_abbr[m] for m in range(1, 13)])
    plt.title(f"Events per Month — {yr}", fontsize=16, fontweight="bold", pad=15)
    plt.xlabel("Month", fontsize=12)
    plt.ylabel("Events", fontsize=12)

    plt.tight_layout()
    plt.show()
```

# For the top 20 counties, what kind of artificial hazards exist in the dataset?

```python
TOP_N_COUNTIES = 20

# --- pick top counties ---
top_counties = (
    acled['admin2']
      .astype(str).str.strip()
      .value_counts()
      .head(TOP_N_COUNTIES)
      .index
)
```

```python
df = acled[acled['admin2'].astype(str).str.strip().isin(top_counties)].copy()
df['admin2'] = df['admin2'].astype(str).str.strip()
```

```python
# crosstab: rows = county, cols = event_type
ct_event = pd.crosstab(df['admin2'], df['event_type']).loc[top_counties]

# nice color palette for event types
colors = ["#4C72B0", "#FF8C42", "#2E8B57", "#D62728",
          "#9467BD", "#8C564B", "#E377C2", "#7F7F7F"]

fig, ax = plt.subplots(figsize=(12, 6))

ct_event.plot(
    kind='bar',
    stacked=True,
    ax=ax,
    color=colors[:len(ct_event.columns)],
    edgecolor="black",
    alpha=0.9,
)

# titles & labels
ax.set_title(f'Event Types in Top {TOP_N_COUNTIES} Counties',
             fontsize=16, fontweight="bold", pad=15)
ax.set_xlabel('County (admin2)', fontsize=12)
ax.set_ylabel('Event count', fontsize=12)

# x labels style
ax.set_xticklabels(ax.get_xticklabels(), rotation=45, ha='right')

# grid + clean frame (same style as before)
ax.grid(axis='y', linestyle='--', alpha=0.4)
for spine in ["top", "right"]:
    ax.spines[spine].set_visible(False)

# legend outside
ax.legend(title='Event type', bbox_to_anchor=(1.02, 1),
          loc='upper left', borderaxespad=0.)

# add total count on top of each stacked bar
totals = ct_event.sum(axis=1).values
for i, total in enumerate(totals):
    ax.text(
        i,
        total,
        f"{int(total):,}",
        ha="center",
        va="bottom",
        fontsize=9,
    )

fig.tight_layout()
plt.show()
```

## What are the value counts of each disorder type?

```python
# Count disorder types
disorder_counts = (
    acled["disorder_type"]
    .dropna()
    .astype(str).str.strip()
    .value_counts()
)

disorder_labels = [f"{d}\n{n:,}" for d, n in disorder_counts.items()]

# Colors from a colormap
cmap = cm.get_cmap("tab20c")
colors = [cmap(i / max(1, len(disorder_counts) - 1)) for i in range(len(disorder_counts))]

plt.figure(figsize=(10, 6))
squarify.plot(
    sizes=disorder_counts.values,
    label=disorder_labels,
    color=colors,
    pad=True,
    text_kwargs={"fontsize": 9},
)

plt.axis("off")
plt.title("Disorder Types by Record Count", fontsize=14, fontweight="bold", pad=10)
plt.tight_layout()
plt.show()
```

## For every disorder type, what are the value counts of each event type?

```python
# Clean columns
tmp = acled[["disorder_type", "event_type"]].copy()
for col in ["disorder_type", "event_type"]:
    tmp[col] = tmp[col].dropna().astype(str).str.strip()

# Loop over each disorder type
for disorder, grp in tmp.groupby("disorder_type"):
    event_counts = grp["event_type"].value_counts()
    if event_counts.empty:
        continue

    labels = [f"{e}\n{n:,}" for e, n in event_counts.items()]

    cmap = cm.get_cmap("viridis")
    colors = [cmap(i / max(1, len(event_counts) - 1)) for i in range(len(event_counts))]

    plt.figure(figsize=(10, 6))
    squarify.plot(
        sizes=event_counts.values,
        label=labels,
        color=colors,
        pad=True,
        text_kwargs={"fontsize": 8},
    )

    plt.axis("off")
    plt.title(f"Event Types within Disorder: {disorder}",
              fontsize=14, fontweight="bold", pad=10)
    plt.tight_layout()
    plt.show()
```

## For each event type, what are the value counts of each sub event type?

```python
# Clean columns
tmp = acled[["event_type", "sub_event_type"]].copy()
for col in ["event_type", "sub_event_type"]:
    tmp[col] = tmp[col].dropna().astype(str).str.strip()

# Loop over each event type
for evt, grp in tmp.groupby("event_type"):
    sub_counts = grp["sub_event_type"].value_counts()
    if sub_counts.empty:
        continue

    # (optional) if there are *tons* of sub-event types, limit to top N
    # N = 20
    # sub_counts = sub_counts.head(N)

    labels = [f"{s}\n{n:,}" for s, n in sub_counts.items()]

    cmap = cm.get_cmap("plasma")
    colors = [cmap(i / max(1, len(sub_counts) - 1)) for i in range(len(sub_counts))]

    plt.figure(figsize=(10, 6))
    squarify.plot(
        sizes=sub_counts.values,
        label=labels,
        color=colors,
        pad=True,
        text_kwargs={"fontsize": 7},
    )

    plt.axis("off")
    plt.title(f"Sub-Event Types within Event: {evt}",
              fontsize=14, fontweight="bold", pad=10)
    plt.tight_layout()
    plt.show()
```

## What disorder type and what event type do these actors belong to?

```python
TOP_N_ACTORS = 20  # change if you want more/less actors

# ---- pick top actors ----
top_actors = (
    acled["actor1"]
      .dropna()
      .astype(str).str.strip()
      .value_counts()
      .head(TOP_N_ACTORS)
      .index
)

# filtered subset
dfA = acled[
    acled["actor1"].astype(str).str.strip().isin(top_actors)
].copy()

for col in ["actor1", "disorder_type", "event_type"]:
    dfA[col] = dfA[col].astype(str).str.strip()

# ==== helper to plot a nice heatmap ====
def plot_actor_heatmap(heat, title):
    plt.figure(figsize=(12, 8))
    ax = plt.gca()

    im = ax.imshow(heat.values, aspect="auto", cmap="viridis")
    cbar = plt.colorbar(im)
    cbar.set_label("Event count", fontsize=12)

    ax.set_yticks(range(len(heat.index)))
    ax.set_yticklabels(heat.index, fontsize=9)

    ax.set_xticks(range(len(heat.columns)))
    ax.set_xticklabels(heat.columns, rotation=45, ha="right", fontsize=9)

    ax.set_title(title, fontsize=16, fontweight="bold", pad=15)

    # clean frame + light grid
    for spine in ["top", "right"]:
        ax.spines[spine].set_visible(False)

    ax.set_xticks(np.arange(-0.5, len(heat.columns), 1), minor=True)
    ax.set_yticks(np.arange(-0.5, len(heat.index), 1), minor=True)
    ax.grid(which="minor", color="white", linestyle="-", linewidth=0.5, alpha=0.7)
    ax.tick_params(which="minor", bottom=False, left=False)

    # annotate cells with counts
    data = heat.values
    max_val = data.max() if data.size else 0
    for i in range(data.shape[0]):
        for j in range(data.shape[1]):
            val = data[i, j]
            if val == 0:
                continue
            color = "white" if max_val and val > max_val * 0.5 else "black"
            ax.text(j, i, int(val), ha="center", va="center", fontsize=7, color=color)

    plt.tight_layout()
    plt.show()


# ==== 1) disorder_type by actor (top actors) ====
heat_disorder = pd.crosstab(dfA["actor1"], dfA["disorder_type"]).loc[top_actors]
plot_actor_heatmap(heat_disorder, f"Disorder Types for Top {TOP_N_ACTORS} Actors")
```

```python
# ==== 2) event_type by actor (top actors) ====
heat_event = pd.crosstab(dfA["actor1"], dfA["event_type"]).loc[top_actors]
plot_actor_heatmap(heat_event, f"Event Types for Top {TOP_N_ACTORS} Actors")
```

## How many records do we have per state?

```python
top_counts = (
    acled['admin1']
      .dropna()
      .astype(str).str.strip()
      .value_counts()
      .head(50)
)

top_labels = [f"{c}\n{n:,}" for c, n in top_counts.items()]

# build a color list from a colormap
cmap = cm.get_cmap("tab20")   # try "plasma", "magma", "tab20" etc.
norm = plt.Normalize(top_counts.min(), top_counts.max())
colors = [cmap(norm(v)) for v in top_counts.values]

plt.figure(figsize=(9, 6))

squarify.plot(
    sizes=top_counts.values,
    label=top_labels,
    color=colors,               # mixed colors 🎨
    pad=True,
    text_kwargs={"fontsize": 9}
)

plt.axis('off')
plt.title('States by Record Count', fontsize=14, fontweight="bold", pad=10)
plt.tight_layout()
plt.show()
```

# This variables seems out of scope as well

```python
# Count fatalities
fatal_counts = (
    acled["fatalities"]
        .fillna(0)
        .astype(int)
        .value_counts()
        .sort_index()
)

# Split into all vs. >0
fatal_counts_pos = fatal_counts[fatal_counts.index > 0]

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# ---- 1) All events (log scale) ----
ax = axes[0]
bars = ax.bar(
    fatal_counts.index.astype(str),
    fatal_counts.values,
    color="#4C72B0",
    edgecolor="black",
    alpha=0.9,
)
ax.set_title("Fatalities per Event (all)", fontsize=14, fontweight="bold", pad=10)
ax.set_xlabel("Fatalities", fontsize=12)
ax.set_ylabel("Number of events", fontsize=12)
ax.set_yscale("log")
ax.grid(axis="y", linestyle="--", alpha=0.4, which="both")
for spine in ["top", "right"]:
    ax.spines[spine].set_visible(False)

# ---- 2) Only events with ≥1 fatality ----
ax = axes[1]
bars2 = ax.bar(
    fatal_counts_pos.index.astype(str),
    fatal_counts_pos.values,
    color="#FF8C42",
    edgecolor="black",
    alpha=0.9,
)
ax.set_title("Fatalities per Event (fatal events only)", fontsize=14, fontweight="bold", pad=10)
ax.set_xlabel("Fatalities", fontsize=12)
ax.set_ylabel("Number of events", fontsize=12)
ax.grid(axis="y", linestyle="--", alpha=0.4)
for spine in ["top", "right"]:
    ax.spines[spine].set_visible(False)

# value labels on the second plot
for b in bars2:
    h = b.get_height()
    ax.text(
        b.get_x() + b.get_width() / 2,
        h,
        f"{int(h):,}",
        ha="center",
        va="bottom",
        fontsize=9,
    )

plt.tight_layout()
plt.show()
```

# Most events have no fatalities

```python

plt.figure(figsize=(8, 6))

plt.scatter(
    acled["longitude"],
    acled["latitude"],
    s=5,
    alpha=0.3,
)

plt.title("Event Locations", fontsize=16, fontweight="bold", pad=15)
plt.xlabel("Longitude", fontsize=12)
plt.ylabel("Latitude", fontsize=12)

ax = plt.gca()
ax.grid(True, linestyle="--", alpha=0.4)

for spine in ["top", "right"]:
    ax.spines[spine].set_visible(False)

ax.set_aspect("equal", adjustable="box")  # keep map-like proportions

plt.tight_layout()
plt.show()
```

```python
# 1) Aggregate events by unique (lon, lat)
coords = (
    acled[["longitude", "latitude"]]
      .dropna()
      .groupby(["longitude", "latitude"])
      .size()
      .reset_index(name="count")
)

# 2) Bigger bubble plot: size + color = count
plt.figure(figsize=(12, 8))  # ⬅️ made the plot bigger

sc = plt.scatter(
    coords["longitude"],
    coords["latitude"],
    s=(coords["count"] / coords["count"].max()) * 300,  # scale size
    c=coords["count"],
    cmap="viridis",
    alpha=0.6,
    edgecolor="black",
    linewidth=0.5,
)

cbar = plt.colorbar(sc)
cbar.set_label("Number of events", fontsize=12)

plt.title("Event Locations (Bubble Map: Size & Color = Count)",
          fontsize=16, fontweight="bold", pad=15)
plt.xlabel("Longitude", fontsize=12)
plt.ylabel("Latitude", fontsize=12)

ax = plt.gca()
ax.grid(True, linestyle="--", alpha=0.3)
for spine in ["top", "right"]:
    ax.spines[spine].set_visible(False)

ax.set_aspect("equal", adjustable="box")

plt.tight_layout()
plt.show()
```
