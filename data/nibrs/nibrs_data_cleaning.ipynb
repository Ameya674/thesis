```python
import pandas as pd
```

# We only keep the ORI, year, state, state_abb, incident date, and ucr_offense_code columns, as they are the only ones required for analysis

```python
cols_keep = [
    'ori',
    'state',
    'incident_date',
    'ucr_offense_code'
]
```

```python
nibrs_2017 = pd.read_csv("nibrs_offense_segment_2017.csv", usecols=cols_keep)
nibrs_2018 = pd.read_csv("nibrs_offense_segment_2018.csv", usecols=cols_keep)
nibrs_2019 = pd.read_csv("nibrs_offense_segment_2019.csv", usecols=cols_keep)
nibrs_2020 = pd.read_csv("nibrs_offense_segment_2020.csv", usecols=cols_keep)
nibrs_2021 = pd.read_csv("nibrs_offense_segment_2021.csv", usecols=cols_keep)
nibrs_2022 = pd.read_csv("nibrs_offense_segment_2022.csv", usecols=cols_keep)
nibrs_2023 = pd.read_csv("nibrs_offense_segment_2023.csv", usecols=cols_keep)
nibrs_2024 = pd.read_csv("nibrs_offense_segment_2024.csv", usecols=cols_keep)
```

## We only keep agencies that were already reporting in 2017 to ensure consistency

```python
# 1. Clean all ORI columns (strip spaces)
for df in [
    nibrs_2017, nibrs_2018, nibrs_2019, nibrs_2020,
    nibrs_2021, nibrs_2022, nibrs_2023, nibrs_2024
]:
    df['ori'] = df['ori'].str.strip()
```

```python
# 2. Get the set of ORIs that are present in 2017
oris_2017 = set(nibrs_2017['ori'].unique())
print("Number of agencies in 2017:", len(oris_2017))
```

```python
# 3. Filter each year to keep only ORIs that existed in 2017
nibrs_2017 = nibrs_2017[nibrs_2017['ori'].isin(oris_2017)].copy()
nibrs_2018 = nibrs_2018[nibrs_2018['ori'].isin(oris_2017)].copy()
nibrs_2019 = nibrs_2019[nibrs_2019['ori'].isin(oris_2017)].copy()
nibrs_2020 = nibrs_2020[nibrs_2020['ori'].isin(oris_2017)].copy()
nibrs_2021 = nibrs_2021[nibrs_2021['ori'].isin(oris_2017)].copy()
nibrs_2022 = nibrs_2022[nibrs_2022['ori'].isin(oris_2017)].copy()
nibrs_2023 = nibrs_2023[nibrs_2023['ori'].isin(oris_2017)].copy()
nibrs_2024 = nibrs_2024[nibrs_2024['ori'].isin(oris_2017)].copy()
```

```python
# merging all individual datasets
nibrs = pd.concat([nibrs_2017, nibrs_2018, nibrs_2019, nibrs_2020, nibrs_2021, nibrs_2022, nibrs_2023, nibrs_2024], ignore_index = True)
```

## These are the 3 categories that the FBI classifies the offenses into, but here property crime also contains bribery, or forgery, or online offenses, that are not physical/hands-on crimes. The crimes marked with a checkbox are the only ones that are physical/hands-on crimes.

## Since we are only looking at property crime, we filter out all other crimes.

```python
physical_property_codes = [
    "arson",
    "destruction/damage/vandalism of property",
    "burglary/breaking and entering",
    "stolen property offenses (receiving, selling, etc.)",
    "robbery",
    "larceny/theft offenses - theft from building",
    "larceny/theft offenses - all other larceny",
    "larceny/theft offenses - shoplifting",
    "larceny/theft offenses - theft from motor vehicle",
    "larceny/theft offenses - pocket-picking",
    "motor vehicle theft",
    "larceny/theft offenses - theft of motor vehicle parts/accessories",
    "larceny/theft offenses - purse-snatching",
    "larceny/theft offenses - theft from coin-operated machine or device",
]
```

```python
nibrs['ucr_offense_code'] = (
    nibrs['ucr_offense_code'].astype(str).str.strip()
)
```

```python
# Keep only rows where the offense is a physical property crime
nibrs = nibrs[
    nibrs['ucr_offense_code'].isin(physical_property_codes)
].copy()
```

```python
print("Physical property rows:", len(nibrs))
print("Unique offense codes kept:", nibrs['ucr_offense_code'].unique())
```
